```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score, roc_auc_score
import joblib
```

```python
# Load train and test datasets
train_df = pd.read_csv("../data/train_data.csv")
test_df = pd.read_csv("../data/test_data.csv")
```

```python
X_train = train_df.drop(columns=["Need_Maintenance"])
y_train = train_df["Need_Maintenance"]

X_test = test_df.drop(columns=["Need_Maintenance"])
y_test = test_df["Need_Maintenance"]
```

```python
# Train the model
rf = RandomForestClassifier(n_estimators=200, max_depth=None, min_samples_split=2, random_state=42)
rf.fit(X_train, y_train)
```

```python
# Predict
y_pred = rf.predict(X_test)
```

```python
# Evaluate
print("\nModel Evaluation:")
print("F1-Score:", f1_score(y_test, y_pred))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_pred))
```

```python
# Save model
joblib.dump(rf, "../models/rf_model.joblib")
print("✅ Model saved to 'models/rf_model.joblib'")
```
